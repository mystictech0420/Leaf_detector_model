# Image feature engineering and CSV export

Create separate feature-enriched `train.csv` and `valid.csv` datasets. The source images remain unchanged and are still used by the CNN loader. Color statistics and histograms are normalized per image; the baseline's scaling parameters are later fitted on training data only.


```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "src").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent
if not (PROJECT_ROOT / "src").is_dir():
    raise FileNotFoundError("Run this notebook from the repository root or notebooks folder.")
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
import csv
from src.feature_engineering import generate_feature_csvs

outputs = generate_feature_csvs()
for split, path in outputs.items():
    with path.open(newline="", encoding="utf-8") as file:
        reader = csv.DictReader(file)
        first_row = next(reader, None)
        print(f"{split}: {path}")
        print(f"Columns ({len(reader.fieldnames or [])}): {reader.fieldnames}")
        print(f"First sample: {first_row}")
```
